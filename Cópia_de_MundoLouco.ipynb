```python

import gradio as gr
import edge_tts
import asyncio
import os
import uuid
import time
from PIL import Image

# Gerar áudio com voz masculina
async def gerar_audio(texto, caminho):
    comunicacao = edge_tts.Communicate(
        texto,
        voice="pt-PT-DuarteNeural",
        rate="+5%"
    )
    await comunicacao.save(caminho)

# MundoLouco: texto + voz + fotografias
def conversar_com_voz(mensagem, history):
    texto = mensagem.get("text", "") or ""
    ficheiros = mensagem.get("files", []) or []

    if not texto and not ficheiros:
        return "Envia uma mensagem ou fotografia.", None

    try:
        if texto:
            guardar_memoria_automatica(texto)

        dados = carregar_memoria()
        memoria = "\n".join(dados.get("informacoes", []))

        prompt = f"""
{INSTRUCOES}

MEMÓRIA DO UTILIZADOR:
{memoria}

MENSAGEM:
{texto if texto else "Analisa esta fotografia e explica o seu conteúdo."}

Responde de forma clara, natural e objectiva.
Se for matéria escolar, explica passo a passo.
"""

        conteudos = [prompt]

        for ficheiro in ficheiros:
            caminho = ficheiro.get("path") if isinstance(ficheiro, dict) else ficheiro

            if caminho and os.path.exists(caminho):
                imagem = Image.open(caminho)
                conteudos.append(imagem)

        resposta_texto = None

        for tentativa in range(3):
            try:
                resposta = cliente.models.generate_content(
                    model=MODELO,
                    contents=conteudos
                )
                resposta_texto = resposta.text or "Não consegui interpretar o conteúdo."
                break

            except Exception:
                if tentativa == 2:
                    return "Não consegui responder. Tenta novamente.", None
                time.sleep(2 ** tentativa)

        # Gerar áudio da resposta
        ficheiro_audio = f"/content/mundolouco_{uuid.uuid4().hex}.mp3"

        try:
            asyncio.run(gerar_audio(resposta_texto, ficheiro_audio))

            if os.path.exists(ficheiro_audio) and os.path.getsize(ficheiro_audio) > 1000:
                return resposta_texto, ficheiro_audio

        except Exception as erro_audio:
            print("Erro na geração de áudio:", erro_audio)

        return resposta_texto, None

    except Exception as erro:
        return f"Erro ao processar: {erro}", None


# Interface com texto, fotografias e áudio
app = gr.ChatInterface(
    fn=conversar_com_voz,
    title="🌍 MUNDOLOUCO",
    description="Assistente inteligente, professor particular e agora com voz!",
    multimodal=True,
    additional_outputs=[
        gr.Audio(label="🔊 Ouvir resposta do MundoLouco", type="filepath")
    ]
)

app.launch(share=True)
```

```python

import unicodedata

def guardar_memoria_automatica(mensagem):
    texto = mensagem.strip()
    normalizado = unicodedata.normalize(
        "NFD", texto.lower()
    )
    normalizado = "".join(
        c for c in normalizado
        if unicodedata.category(c) != "Mn"
    )

    indicadores = [
        "o meu objectivo e",
        "meu objetivo e",
        "gosto de",
        "prefiro",
        "estou a estudar",
        "trabalho como",
        "o meu projecto e",
        "meu projeto e",
        "sou estudante",
        "quero desenvolver"
    ]

    if not any(p in normalizado for p in indicadores):
        return False

    dados = carregar_memoria()
    informacoes = dados.setdefault("informacoes", [])

    if texto not in informacoes:
        informacoes.append(texto)
        guardar_memoria(dados)
        return True

    return False

print("🧠 Memória automática preparada!")
```
